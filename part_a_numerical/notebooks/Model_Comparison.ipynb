```python
import pandas as pd

lr = pd.read_csv('../results/linear_regression_results.csv')
knn = pd.read_csv('../results/knn_results.csv')

comparison = pd.concat([lr, knn], ignore_index=True)

comparison
```

```python
import matplotlib.pyplot as plt

comparison.set_index('Model')[['MAE', 'RMSE']].plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Model Comparison - MAE & RMSE')
plt.ylabel('Error')
plt.xticks(rotation=0)
plt.show()
```

```python
comparison.plot(
    x='Model',
    y='R2 Score',
    kind='bar',
    figsize=(8, 5),
    legend=False
)

plt.title('R2 Score Comparison')
plt.ylabel('R2 Score')
plt.xticks(rotation=0)
plt.show()
```
